```python
EPOCHS = 50
IMAGE_SIZE = 640
BATCH_SIZE = 16

PROJECT = "/content/drive/MyDrive/Major_Project_AI/models"

RUN_NAME = "YOLO11_RSNA_v1"

DEVICE = 0

PATIENCE = 15

WORKERS = 2

PRETRAINED = True
```

# Training the YOLOv11 Model

In this stage, we train a pretrained YOLOv11 Nano model using transfer learning on the RSNA Pneumonia Detection dataset.

The objective is to learn to detect pneumonia regions in chest X-ray images by predicting bounding boxes around infected areas.

## Training Configuration

- Model: YOLO11 Nano
- Learning Method: Transfer Learning
- Epochs: 50
- Image Size: 640 × 640
- Batch Size: 16
- Optimizer: Auto (selected by Ultralytics)
- Device: NVIDIA Tesla T4 GPU
- Early Stopping: Enabled

```python
results = model.train(
    data=yaml_path,
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,
    project=PROJECT,
    name=RUN_NAME,

    device=DEVICE,
    workers=WORKERS,
    patience=PATIENCE,
    pretrained=PRETRAINED,

    optimizer="auto",

    cache=False,
    save=True,
    save_period=-1,
    plots=True,
    verbose=True,

    val=True
)
```
